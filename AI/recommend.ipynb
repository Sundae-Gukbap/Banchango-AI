```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# 레시피 데이터 읽기
recipe_df = pd.read_csv("Last_recipe.csv")

# TF-IDF 벡터라이저 생성
tfidf = TfidfVectorizer(token_pattern=r'[^|]+')
tfidf_matrix = tfidf.fit_transform(recipe_df['Ingr2'])

user_ingredients = ['소금', '닭가슴살', '간장', '설탕', '계란', '콩나물', '양파', '마늘']
user_ingredients_str = '|'.join(user_ingredients)
print(user_ingredients_str)
user_vector = tfidf.transform([user_ingredients_str])

scores = cosine_similarity(tfidf_matrix, user_vector).flatten()

# 상위 N개 레시피 추천
N = 20
top_indices = scores.argsort()[-N:]
top_recipes = recipe_df.iloc[top_indices]

for i, row in top_recipes.iterrows():
    print(f"추천 레시피: {row['best_name']} (점수: {scores[i]}), {row['index']}, {row['name']}")


# 장점: 1. 설명 가능성: TF-IDF는 단어의 중요도를 직관적으로 파악할 수 있어, 추천 결과에 대한 설명이 용이합니다.
#      2. 새로운 재료에 대한 대응: 새로운 재료가 추가되더라도 TF-IDF 점수를 계산하여 반영할 수 있습니다.
#      3. 콘텐츠 기반 추천: 재료 간 유사도를 기반으로 추천하므로, 사용자의 과거 행동 데이터가 없어도 추천이 가능합니다.

#단점: 1. 정확도 한계: TF-IDF는 단순히 재료의 중요도를 계산하므로, 사용자의 선호도나 맥락을 충분히 반영하기 어려워 추천 정확도가 낮을 수 있습니다.
#     2. 다양성 부족: 재료의 유사도만을 고려하므로, 다양한 재료를 추천하기 어려워 사용자의 선택 범위가 좁아질 수 있습니다.
#     3. 콘텐츠 기반 한계: 재료 간 유사도만을 활용하므로, 사용자의 행동 데이터를 활용한 협업 필터링 기법에 비해 성능이 떨어질 수 있습니다.
#     4. 사용자 선호도 반영 어려움: 사용자의 개인적인 선호도나 취향을 반영하기 어려워, 개인화된 추천이 어려울 수 있습니다.

# 추후 단점을 보안한 개인화된 추천이 될 수 있도록 노력하겠습니다.
```
