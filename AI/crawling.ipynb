```python
import requests, json, csv
from requests.adapters import HTTPAdapter
from requests.packages.urllib3.util.retry import Retry

# 재시도 전략 설정
retries = Retry(total=10,  # 최대 재시도 횟수
                backoff_factor=1,  # 지연 시간에 대한 백오프 인자
                status_forcelist=[500, 502, 503, 504, 104, 10054, 2])  # 재시도할 상태 코드

# HTTPAdapter와 재시도 전략을 사용해 Session 객체 생성
session = requests.Session()
adapter = HTTPAdapter(max_retries=retries)
session.mount('http://', adapter)
session.mount('https://', adapter)
```

```python
import pandas as pd
df = pd.read_csv('recipe.csv')
```

```python
from konlpy.tag import Okt
from tqdm import tqdm

okt = Okt()

df['Ingr'] = None

for index, row in tqdm(df.iterrows()):
    x = okt.nouns(row['recipeIngredient'])
    df.at[index, 'Ingr'] = x
```

```python
all_ingredients = sum(df['Ingr'].tolist(), [])
print('1')
# 중복 제거
unique_ingredients = set(all_ingredients)
print('2')
# 재료 목록을 텍스트 파일로 저장
with open('ingredients.txt', 'w', encoding='utf-8') as file:
    for ingredient in sorted(unique_ingredients):
        file.write(f"{ingredient}\n")

print('재료 목록이 ingredients.txt 파일에 저장되었습니다.')
```
